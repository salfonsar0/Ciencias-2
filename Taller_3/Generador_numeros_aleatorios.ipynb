# Generador de numeros aleatorios

Desarrollar un algoritmo generador de números pseudoaleatorios con distribución uniforme para un espacio muestral de 20 valores (0 al 19), utilizando el método de Generador Congruencial Lineal (LCG). El diseño asegura la estabilidad matemática y previene el colapso temprano del ciclo iterativo, permitiendo la evaluación de su distribución estadística.

## 1. Definicion de variables
El algoritmo se fundamenta en la relación de recurrencia clásica del LCG, manteniendo el cálculo interno en un espacio numérico amplio antes de proyectarlo al espacio reducido deseado. A continuación, se detalla la función de cada variable dentro del sistema:

*   **semilla_inicial**: Valor numérico de entrada proporcionado. Actúa como el vector de inicialización (estado cero) del sistema generador.
*   **modulo**, **multiplicador** e **incremento**: Constantes matemáticas paramétricas del algoritmo. Se emplean valores estandarizados para garantizar la máxima longitud del ciclo y mantener la uniformidad en la dispersión de los datos.
*   **estado_actual**: Variable de estado interno que preserva la memoria de la secuencia. Al iterar sobre un módulo extenso ($2^{31}$), se evita el estancamiento del algoritmo en bucles cortos.
*   **frecuencias**: Estructura de datos tipo lista de 20 posiciones, empleada como arreglo de contadores para el registro y análisis estadístico.
*   **iteracion**: Variable de control para el ciclo principal de procesamiento, establecido en 100,000 ciclos para generar una muestra estadísticamente significativa.
*   **valor_final**: Resultado de aplicar la operación módulo 20 al estado interno, garantizando que los resultados de salida se restrinjan estrictamente a los límites requeridos (0 a 19).
*   **numero**: Variable de iteración utilizada en la fase de comprobación analítica para recorrer la estructura de frecuencias.
## 2. Codigo

```python
semilla_inicial = int(input("Ingresa la semilla: "))

frecuencias = [0] * 20
modulo = 2147483648
multiplicador = 1103515245
incremento = 12345
estado_actual = semilla_inicial

print("\nPrimeros 10 números generados:")

for iteracion in range(100000):
    estado_actual = (multiplicador * estado_actual + incremento) % modulo
    valor_final = estado_actual % 20

    frecuencias[valor_final] += 1

    if iteracion < 10:
        print(valor_final, end=" ")

print("\n\n=== VERIFICACIÓN DE DISTRIBUCIÓN ===")
print("En un algoritmo perfecto, cada número debería salir unas 5000 veces.")

for numero in range(20):
    print(f"Número {numero}: salió {frecuencias[numero]} veces")
```

## 3. Análisis y Verificación Estadística
El sistema incorpora una rutina de validación que procesa 100,000 números. Bajo una distribución uniforme ideal, la probabilidad marginal de aparición de cada valor dentro de un espacio de 20 elementos es de $1/20$. Por consiguiente, la esperanza matemática para cada número es de aproximadamente 5,000 ocurrencias. La última sección del código expone estos resultados tabulados para validar empíricamente la calidad y uniformidad del generador implementado
